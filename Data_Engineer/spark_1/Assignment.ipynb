# Spark Assignment

Based on the movie review dataset in 'netflix-rotten-tomatoes-metacritic-imdb.csv', answer the below questions.

**Note:** do not clean or remove missing data

```python
path = "netflix-rotten-tomatoes-metacritic-imdb.csv"
df = spark.read.csv(path, header=True, inferSchema=True)
print(df.printSchema())
```

```python
cols = [col.replace(" ","_") for col in df.columns]
df = df.toDF(*cols)
print(df.printSchema())
```

## What is the maximum and average of the overall hidden gem score?

```python
results = df.select(max("Hidden_Gem_Score"), avg("Hidden_Gem_Score")).collect()
print(f"max: {round(results[0][0],2)}, avg: {round(results[0][1],2)}")
```

## How many movies that are available in Korea?

```python
results = df.filter(df["Languages"].contains("Korea")).select("Title").distinct().count()
print(results)
```

## Which director has the highest average hidden gem score?

```python
results = df.groupBy("Director").agg({"Hidden_Gem_Score":"avg"}).orderBy("avg(Hidden_Gem_Score)", ascending=False).collect()
print(results[0][0])
```

## How many genres are there in the dataset?

```python
df = df.filter(df["Genre"].isNotNull())
results = results = df.select("Genre").rdd.flatMap(lambda x: x[0].split(",")).map(lambda x: x.strip()).map(lambda x: (x, 1)).reduceByKey(lambda x, y: x + y).collect()
print(results)
```
