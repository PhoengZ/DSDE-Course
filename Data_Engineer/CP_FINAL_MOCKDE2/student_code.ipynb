# Q1: Web Scraping – Count National Holidays by Weekday

You are provided with an HTML file containing Thailand’s national holidays for a given year.

Your task is to extract and analyze the holiday data from the HTML file.

- Load the HTML content from the given `file_path`.
- Extract all national holiday dates from the HTML.
- Determine the weekday (Monday–Sunday) for each holiday.
- Count how many holidays fall on each weekday.
- Store the counts in a list ordered from Monday (index 0) to Sunday (index 6) as `holiday_counts`.

Output [Grader: Yes]: Return the dictionary containing:  
```python
{
    "holiday_counts": <list of counts from Monday to Sunday>
}
```

Note: The result must be a list of 7 integers.  

````python
with open(file_path, "r", encoding="utf-8") as file:
    html_content = file.read()
soup = BeautifulSoup(html_content, "lxml")
trs = soup.select("tr.country")
d = {
    "Monday":0,
    "Tuesday":1,
    "Wednesday":2,
    "Thursday":3,
    "Friday":4,
    "Saturday":5,
    "Sunday":6
}
results = [0 for i in range(7)]
for tr in trs:
    if "National Holiday" in tr.text.strip():
        td = tr.select_one("td")
        results[d[td.text.strip()]]+=1
print(results)
# Write your code here
````

````python
import pandas as pd
with open(file_path, "r", encoding="utf-8") as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, "lxml")
trs = soup.select("tr.country")
results = [0 for i in range(7)]
for tr in trs:
    if "National Holiday" in tr.text.strip():
        time = tr.select_one("time")
        dt = time.attrs['datetime']
        t = pd.to_datetime(dt, format="%Y-%m-%d")
        results[t.day_of_week]+=1
print(results)
````

# Q2: Web Scraping – Find Visakha Bucha Day

You are provided with an HTML file containing Thailand’s national holidays for a given year.

Your task is to locate and extract information about **Visakha Bucha Day**.

- Load the HTML content from the given `file_path`.
- Search for the entry corresponding to "Visakha Bucha Day".
- Extract the date of this holiday.
- Determine the weekday (e.g., Monday, Tuesday, etc.).
- Store the result as `visakha_bucha_day`.

Output [Grader: Yes]: Return the dictionary containing:  
```python
{
    "visakha_bucha_day": <weekday of Visakha Bucha Day>
}
```

Note: The output should be a string representing the weekday.  
Hint: Look for the holiday name in the HTML, then convert its date into a weekday.

````python
with open(file_path, "r", encoding="utf-8") as file:
    html_content = file.read()
soup = BeautifulSoup(html_content, "lxml")
trs = soup.select("tr.country")
for tr in trs:
    if "Visakha Bucha Day" in tr.text.strip():
        td = tr.select_one("td")
        print(td.text.strip())
        break
else:
    print("Not found")
# Write your code here
````
