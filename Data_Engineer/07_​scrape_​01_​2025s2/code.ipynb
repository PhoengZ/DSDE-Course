```python
def Q2(file_path):
    with open(file_path, "r") as f:
        html = f.read()
    soup = BeautifulSoup(html, "lxml")
    all_div = soup.select("div.bud-day-col")
    for i, e in enumerate(all_div):
        a = e.find("a")
        if a and "วันวิสาขบูชา" in a.text.strip():
            return all_div[i].parent.parent.find_next()
    return None
```

```python
print(Q2("2566.html"))
```

```python
def Q1(file_path):
    with open(file_path, "r")as f:
        html = f.read()
    soup = BeautifulSoup(html, "lxml")
    divs = soup.select("div.bud-day-col")
    days = [div for div in divs if "2566" in div.text.strip()]
    check = ["วันจันทร์", "วันอังคาร", "วันพุธ", "วันพฤหัสบดี", "วันศุกร์", "วันเสาร์", "วันอาทิตย์"]
    result = [0 for i in range(7)]
    for day in days:
        for i, c in enumerate(check):
            if c in day.text.strip():
                result[i]+=1
    return result
```

```python
print(Q1("2566.html"))
```
